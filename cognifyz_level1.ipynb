```python
# Task 3: Price Range Distribution

# Count restaurants in each price range
price_count = df['Price range'].value_counts()

# Calculate percentage
price_percentage = (price_count / len(df)) * 100

print(price_percentage)

# Bar chart
import matplotlib.pyplot as plt

plt.bar(price_percentage.index, price_percentage.values)

plt.xlabel("Price Range")
plt.ylabel("Percentage of Restaurants")
plt.title("Price Range Distribution")

plt.savefig('price_range_distribution.png', dpi=300, bbox_inches='tight')

plt.show()
```

```python
import pandas as pd

df = pd.read_csv("Dataset .csv")
# Percentage of restaurants offering online delivery

online_delivery = (df['Has Online delivery'].value_counts(normalize=True) * 100).round(2)

print(online_delivery)
```

```python
# Average rating based on online delivery

avg_rating = df.groupby('Has Online delivery')['Aggregate rating'].mean().round(2)

print(avg_rating)
```

```python
# Distribution of Aggregate Ratings

rating_distribution = df['Aggregate rating'].value_counts().sort_index()

print(rating_distribution)
```

```python
# Average number of votes received

average_votes = df['Votes'].mean()

print("Average number of votes:", round(average_votes, 2))
```

```python
# Find the most common cuisine combinations

cuisine_combination = df['Cuisines'].value_counts().head(10)

print(cuisine_combination)
```

```python
# Display latitude and longitude of restaurants

location_data = df[['Latitude', 'Longitude']]

print(location_data.head())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(5,4))
plt.scatter(df['Longitude'], df['Latitude'], alpha=0.5)

plt.title("Geographic Distribution of Restaurants")
plt.xlabel("Longitude")
plt.ylabel("Latitude")

plt.show()
```

```python
# Top restaurant chains

restaurant_chains = df['Restaurant Name'].value_counts()

print(restaurant_chains.head(10))
```

```python
# Find restaurant chains

chains = df['Restaurant Name'].value_counts()

print(chains.head(10))
```

```python
# Average rating of top restaurant chains

top_chains = df['Restaurant Name'].value_counts().head(10).index

ratings = df[df['Restaurant Name'].isin(top_chains)]

print(ratings.groupby('Restaurant Name')['Aggregate rating'].mean())
```
